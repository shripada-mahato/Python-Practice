Python *Args

```python
def calculate_sum(*args):
    total = 0 
    for i in args:
        total += i
    return total
calculate_sum(10, 20, 30)
```

```python
def count_even(*args):
    count = 0
    for i in args:
        if i % 2 == 0:
            count += 1
    return count
count_even(10, 15, 20, 7, 8, 11)
```

```python
def find_largest(*args):
    largest = args[0]
    for i in args:
        if i > largest:
            largest = i
    return largest
find_largest(-10, -5, -20)
```

```python
def find_smallest(*args):
    smallest = args[0]
    for i in args:
        if i < smallest:
            smallest = i
    return smallest
find_smallest(15, 8, 20, 3, 10)
```

```python
def count_positive(*args):
    positive_count = 0
    for i in args:
        if i > 0:
            positive_count += 1
    return positive_count
count_positive(5, -2, 10, 0, -7, 8)
```

```python
def analyze_number(*args):
    positive = 0
    nagetive = 0
    zero = 0
    for i in args:
        if i > 0:
            positive += 1
        elif i < 0:
            nagetive += 1
        else:
            zero += 1
    return positive, nagetive, zero
analyze_number(5, -2, 0, 8, -7, 0, 3)
```

```python
def calculate_average(*args):
    no = len(args)
    total = 0
    for i in args:
        total += i
    return total / no
calculate_average(10, 20, 30, 40)
```

```python
def multiply_all(*args):
    total = 1
    for i in args:
        total *= i
    return total
multiply_all(2, 3, 4)
```

```python
def count_negative(*args):
    count = 0
    for i in args:
        if i < 0:
            count += 1
    return count
count_negative(5, -2, -8, 10, -3)
```

```python
def find_first_negative(*args):
    num = 0
    for i in range(len(args)):
        if args[i] < 0:
            num = args[i]
            break
    return num
find_first_negative(5, 8, -3, -10, 7)
```

```python
def find_last_negative(*args):
    num = 0
    for i in range(len(args)):
        if args[i] < 0:
            num = args[i]
    return num
find_last_negative(5, 8, -3, -10, -7)
```
