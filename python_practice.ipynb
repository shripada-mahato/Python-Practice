```python
def odd_even(num):
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")
odd_even(10)
```

exception handling

```python
try:
    num = 10 / 2
except ZeroDivisionError:
    print(f"can't devide by zero")
else:
    print(int(num))
```

```python
try:
    print(10/2)
    print(10%2)
except ZeroDivisionError:
    print('error')
finally:
    print('program finished')
```

```python
try:
    age = int(input("enter number:- "))
    print(age)
except ValueError:
    print('invalid number')
```

```python
try:
    num = int(input("enter a number:- "))
    print(num % 0)
except ZeroDivisionError:
    print("invalid number")
```
