```python
def odd_even(num):
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")
odd_even(10)
```

exception handling

```python
try:
    num = 10 / 2
except ZeroDivisionError:
    print(f"can't devide by zero")
else:
    print(int(num))
```

```python
try:
    print(10/2)
    print(10%2)
except ZeroDivisionError:
    print('error')
finally:
    print('program finished')
```

```python
try:
    age = int(input("enter number:- "))
    print(age)
except ValueError:
    print('invalid number')
```

args function

```python
def calculate_sum(*args):
    total = 0
    for i in args:
        total += i
    return total
print(calculate_sum(10,20,30))
```
