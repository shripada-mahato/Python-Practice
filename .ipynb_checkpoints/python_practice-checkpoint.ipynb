args function

```python
def calculate_sum(*args):
    total = 0
    for i in args:
        total += i
    return total
print(calculate_sum(10,20,30))
```

```python
def fnd_largest(*args):
    largest_value = 0
    for i in args:
        if i > largest_value:
            largest_value = i
    return largest_value
fnd_largest(-10, -20, -5, -30)
```

```python
def find_even_numbers(*args):
    even_no = []
    for i in args:
        if i % 2 == 0:
            even_no.append(i)
    return even_no
find_even_numbers(10, 15, 22, 7, 30, 41,42)
```

Challenge 1 — **kwargs

```python
def show_student(**kwargs):
    for key,value in kwargs.items():
        print(key,value)
show_student(name="shri",age=21,course="BCA")
```

```python
def average_marks(**kwargs):
    total = 0
    count = 0
    for value in kwargs.values():
        total += value
        count += 1
    average = total / count
    return average
average_marks(python=81,java=70,sql=90)
```

```python
def calculate_sum(*args):
    sum = 1
    for i in args:
        sum = sum * i
    print(sum)
calculate_sum(2,4,6)
```

```python
def positive_count(*args):
    count = 0
    for i in args:
        if i > 0:
            count += 1
    return count
print(positive_count(10, -5, 20, -3, 0, 15))
```

```python
def analyze_number(*args):
    positive_value = 0
    negative_value = 0
    zero_value = 0
    for i in args:
        if i > 0:
            positive_value += 1
        elif i < 0:
            negative_value += 1
        else:
            zero_value += 1
    return positive_value,negative_value,zero_value
print(analyze_number(10, -5, 0, 20, -3, 0, 15))
```

```python
def calculate_sum(num1):
    sum1 = 0
    for i in range(1, num1+1):
        sum1 += i
    return sum1
calculate_sum(4)
```

```python
def sum_of_even(num):
    sum = 0
    for i in range(1, num+1):
        if i % 2 == 0:
            sum += i
    return sum
sum_of_even(10)
```

```python
def count_vowel(word):
    count = 0
    for i in word:
        if i in 'aeiou':
            count += 1
    return count
count_vowel('hello')
```

```python
def count_consonents(word):
    count = 0
    for i in word:
        if i.lower() not in 'aeiou' and i.isalpha():
            count += 1
    return count 
print(count_consonents('hello 123k'))
```

```python
def find_largest(arr):
    largest_no = arr[0]
    for i in range(len(arr)):
        if arr[i] > largest_no:
            largest_no = arr[i]
    return largest_no
find_largest([10, 31, 25, 8, 15])
```

```python
def count_even(arr):
    count = 0
    for i in arr:
        if i % 2 == 0:
            count += 1
    return count
count_even([10, 3, 8, 7, 12, 5])
```

```python
def analyze_numbers(arr):
    positive = 0
    nagetive = 0
    zero = 0
    for i in arr:
        if i > 0:
            positive += 1
        elif i < 0:
            nagetive += 1
        else:
            zero += 1
    return positive, nagetive, zero
analyze_numbers([5, -2, 0, 8, -7, 0, 3, -4, 7])
```
