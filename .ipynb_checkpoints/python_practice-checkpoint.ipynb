args function

```python
def calculate_sum(*args):
    total = 0
    for i in args:
        total += i
    return total
print(calculate_sum(10,20,30))
```

```python
def fnd_largest(*args):
    largest_value = 0
    for i in args:
        if i > largest_value:
            largest_value = i
    return largest_value
fnd_largest(-10, -20, -5, -30)
```

```python
def find_even_numbers(*args):
    even_no = []
    for i in args:
        if i % 2 == 0:
            even_no.append(i)
    return even_no
find_even_numbers(10, 15, 22, 7, 30, 41,42)
```

```python
n = int(input())
if n % 2 == 1:
    print("Weird")
elif n % 2 == 0 and 2 <= n <= 5:
    print('Not Weird')
elif n % 2 == 0 and 6 <= n <= 20 :
    print('Weird')
else:
    print('Not weird')
```

Challenge 1 — **kwargs

```python
def show_student(**kwargs):
    for key,value in kwargs.items():
        print(key,value)
show_student(name="shri",age=21,course="BCA")
```

```python
def average_marks(**kwargs):
    total = 0
    count = 0
    for value in kwargs.values():
        total += value
        count += 1
    average = total / count
    return average
average_marks(python=81,java=70,sql=90)
```
